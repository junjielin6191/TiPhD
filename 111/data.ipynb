```python
import pandas as pd

# 读取数据
data = pd.read_csv("/mnt/data/ljj/Project_TiPhD/111/data/spatialayer.csv")

# 统计种类数量
cancer_type_unique_count = data['cancer_type'].nunique()
cell_name_unique_count = data['spatial_layer'].nunique()
paper_unique_count = data['PMID'].nunique()
phenotype_label_unique_count = data['Phenotype_type'].nunique()
# 打印结果
print(f"Cancer Type 种类总数: {cancer_type_unique_count}")
print(f"Cell Name 种类总数: {cell_name_unique_count}")
print(f"paper 种类总数: {paper_unique_count}")
print(f"Phenotype Label 种类总数: {phenotype_label_unique_count}")
```

```python
# 输出每个标签的计数
phenotype_label_counts = data['Phenotype_type'].value_counts()

# 打印结果
print(phenotype_label_counts)
```

```python
summary = {
    "Feature": ["Cancer Type", "Cell Name", "Phenotype Label"],
    "Unique Count": [cancer_type_unique_count, cell_name_unique_count, phenotype_label_unique_count]
}

summary_df = pd.DataFrame(summary)
print(summary_df)

# 保存为 CSV 文件
summary_df.to_csv("unique_cell_counts_summary.csv", index=False)
```
